# 이상징후
이상징후는 탐지해줘야 한다..
예: 관제실 모니터에 오늘 부산항 입항척수가 40척이다. 평소에는 120척 정도 들어왔는데 40척이 들어왔다. 아 이거 부산항에 뭔 일 있구나. -> 태풍, 관제 시스템이 먹통이 되어서 기록이 누락되었거나.. 평소보다 조금 들어온 건(100-110척 정도-안정적인 범위 안) 괜찮음.

평소와 다르다를 판단하려면 평소가 무엇인지를 결정해야 하고 얼마나 달라야 이상한 것인지를 결정할 수 있어야 한다.

데이터에서 평소와 다른 값을 찾아서 담당자에게 알리는 이상징후 탐지anomaly detection를 살펴보겠다.

# 이상징후란
이상치outlier 다른 값들과 동떨어져 있는 값을 가리키는 통계 용어이다.
이상징후는 평소의 패턴에서 벗어나서 확인이 필요한 신호를 의미하는데 모든 이상치는 이상징후가 아니다.

20만톤급 컨테이너선의 총톤수는 다른 배들과 다소 떨어져 있는 값일 수 있지만 그런 이상치로 구분될 수 있는 값이 들어왔다고 해서 무조건적으로 이상징후로 받아들일 필요는 없다.

그런데 평균 120척 들어왔는데 원래는 목요일에는 80척 정도가 들어왔던 상황에서 갑자기 목요일에 120척이 들어오면 이상 징후로 볼 수 있다.

이상징후가 발견이 될 때 원인을 구분해보면 크게 세가지로 구분해 볼 수 있다.
- 데이터 오류: ex 3200년 출항이 갑자기 잡힘, 환적 톤수가 20억, 중복기록 -> 고칠 수 있음..
- 운영사건: 장비 고장, 환경문제, 명절
- 구조변화: 신항 개장, 코로나, 선사 파산 - 기준선을 다시 잡아줘야 한다

## 이상의 모양
이상 징후는 모양에 따라 세 가지로 구분해 볼 수 있다.
- 점 이상: 값 하나가 전체에서 동떨어진 경우
- 맥락 이상: 값 자체는 흔한데 그 상황에서는 이상한 것 ex . 110척 들어왔는데 그 요일에는 그렇게 안 들어왔는데..이러면 맥락 이상
- 집단 이상: 값 하나하나는 평범한데 여러 개가 모인 모양이 드문 것
