# 스마트 항만
스마트 항만(smart port)은 항만에서 일어나는 입출항, 하역, 보관, 반출입과 같은 작업을 정보통신 기술로 연결하고 누적된 데이터를 바탕으로 더 빠르고 안전하며 예측 가능하게 운영하는 항만을 의미한다.

스마트폰이 전화기에 인터넷과 센서와 앱을 더해서 전혀 다른 물건이 된 것과 같이 스마트항만도 부두와 크레인이라는 물리적인 사실은 그대로 두되 그 위에 정보 흐름과 자동 판단과 같은 것들을 얹어서 일하는 방식을 바꾼 것으로 볼 수 있다.

정보연결: port-mis, 전자신고, 항만 커뮤니티 시스템
운영 자동화: 원격 안벽 크레인, 무인 이송장비, 자동화 야드 크레인
예측 자동화: 디지털 트윈, 선석 계획 최적화, 도착 예측

## 항만 물류 흐름
- 해측(sea side): 선박의 흐름
    - 입항신고 - 항로 관제 - 도선/예선
    - 묘박지(앵커링을 하는 곳 - 양묘(닻올림)/투묘(닻내림)) 대기
    - 접안 - (크레인을 통해 양하 작업) 하역 - 이안(접안의 반대) - 출항
- 터미널(yard): 컨테이너의 흐름
    - 안벽크레인 양하/적하
    - 이송장비 장치장(cy, container yard) 운반
    - 장치장에 장치해두고 반출 대기
- 육측(내륙): 철도.트럭(애륙운송)
    - 반출입 예약 - 터미널 게이트 통과
    - 장치장 컨테이너 상/하차

각각의 단계는 선행 단계가 완료되어야 시작할 수 있기 때문에 한 단계의 지연은 뒤로 전파될 수 있다. 공선석이 없을 때에는 배는 묘박지에서 대기해야만 하고 장치장 가득차면 크리엔이 컨테이너를 내려놓을 곳이 없어-하역이 느려짐

이 단계들 사이에서 정보 대기시간과 작업 대기시간을 줄이는데 초첨을 맞추는 것이 바람직하다.

### 항만
항구분 - 부두 - 선석
- 항구분 : 항만 안의 큰 권역(청코드, 항구분) 예: 북항, 신항, 감천항(+선박수리도함)
- 부두(terminal) : 한 운영사가 관리하는 하역 구역 예: 신선대부두, 신항 2부두
- 선석(berth):  배 한척이 붙는 자리 - 사용승인 필요 및 종료 신고도 필요 예: 신항 2부두 3선석, 선석 구분코드, 계선장소시설코드/계선장소명(선석)
- 박지(anchorage):  해상 정박 구역 예: 남외항 묘박지, 시설 구분

선석을 주차장으로 비유했을 때 부두는 주차장 한 구역, 선석은 주차 한 칸, 작은 배 2척이 하나의 선석에 댈 수도 있고 초대형 컨테이너선 한 척이 두개의 선석을 차지할 수도 있다. 선석 점유율 계산시 한 선석에 한 척이라는 가정을 하면 오류가 있을 수 있다..

묘박지는 선박이 닻을 내리고 대기하는 장소를 의미한다. 공선석이 생기는 것을 대기하는 선박, 급유, 선원 교대처럼 부두에 붙을 필요가 없는 일을 하는 선박, 태풍을 피하는 선박들이 여기서 대기할 수 있다.

무조건적으로 묘박지 이용(입항)을 양하를 위한 대기로 보면 안되고 입항과 목적을 같이 봐야한다.

## 화물톤, 운임톤(r/t)
컨테이너에 담길 수 없는 물건들(석유, 곡물, 철강 ...) TEU로 헤어릴 수 없기 때문에 톤으로 헤아립니다. 근데 무게만으로 요금을 측정하면 스티로폼과 같이 부피가 큰 화물은 공간은 많이 차지하는데 요금이 거의 없다.

그렇기 때문에 해운업꼐에서는 무게(톤), 부피($m^3$) 각각 계산한 다음에 더 큰 쪽을 기준으로 요금을 산정(운임톤)

### 선박 크기 - 총톤수량 재화중량톤
선박의 크기에서 이야기하는 톤(t)

총톤수(g/t, gross tonnage)는 배의 무게가 아니라 배 내부 공간의 크기(용적)을 환산한 값.
    아파트 집크기 전용면적으로 보는 것과 유사하다
    항만 사용료, 도선/예선 기준, 선박 크기 분류에 사용된다.

재화중량톤(DWT, deadweight tonnage) 배가 안전하게 실을 수 있는 화물/연료/물자의 최대 무게, 벌크나 유조선의 크기를 말할 때 주로 사용됨.

총톤수와 재화중량톤은 선박의 종류에 따라 달라지는데 원유처럼 무거운 화물을 싣는 유조선은 부피에 비해 무거운 물건(기름)을 싣기 때문에 dwt보다 gt가 훨씬 크다. 반면에 자동차 여객선 같이 가볍고 부피가 큰 화물을 싣는 배는 dwt보다 gt가 더 작다.

10만톤급 배면 무조건적으로 엄청 규모가 큰 배다 라고 생각할 것이 아니라 gt인지 gwt인지를 확인해서 판별해야 된다.


수출입 환적 구분
- OO, II, IT, OT
- 수출: 우리나라에서 만들어져서 외국으로 나가는 화물
- 수입: 외국에서 들어와 국내에서 쓰이는 화물
- 수출환적: 환적 화물이 부산에서 다시 배에 실려 나가는 것
- 수입환적: 환적 화물이 부산에서 배에서 내려지는 쪽

환적에서 분석시 문제가 되는 부분 환적 컨테이너 한개가 물동량 통계에서 두번 잡힐 수 있다. 환적 컨테이너는 배에서 한번 내리고(수입환적) 다른배에 한번 싣는다(수출환적) 크레인이 두번 움직이고 항만 처리량에도 두번 더해질 수 있다.

처리량을 항만이 한 일의 양으로 보면 맞는 계산일 수 있는데 부산을 거쳐간 컨테이너 갯수로 읽으면 2배로 부풀려진 지표일 수 있다.

- 항만 처리량(수행한 업무량) = 수출+수입+수출환적+수입환적
- 부산을 거친 컨테이너 수 = 수출+수입+(수출환적+수입환적)/2

## 식별체계 항망 주체
데이터를 서로 연결하기 위해서는 이 기록과 저 기록의 배가 같은 배인지 같은 입항인지를 판단할 수 있어야 한다. 이때 식별자가 필요하다

### 선박의 입항 식별에 사용되는 식별자
- 선박명: 사람이 읽는 배 이름, 배가 매각되면 바뀔 수 있음
- 호출부호: 무선 호출용, 선박의 국적이 바뀌면 바뀔 수 있음<출항 내역 간단히 볼 때>
- IMO: 국제선박등록번호, 바뀌지 않음<모든 내역 확인 시>
- 연도+ 항차: 한 선박의 그 해 입항한 횟수
- UN/LOCODE: 항구코드

### 항만의 주체의 만들어내는 데이터
항만에서는 서로 다른 목적을 가진 여러 주체가 함께 작업을 수행한다. 각자의 목표별로 산출해내는/기록해내는 데이터가 다르다.

- 배를 운항하는 주체: 선사
- 선석과 크레인을 운영: 터미널 운영사
- 화물 주인 : 화주(하주)
- 운송 설계 및 관리: 포워더
- 항만 관리 : 항만청(부산항만공사)
- 관제랑 허가 업무 수행: 해양수산청

같은 입항건이더라도 관심있는 숫자가 다르다.

- 선사: 체류시간, 정시성, 항비
- 선사 대리자(신고업무): 신고/허가 처리 시각
- 터미널 운영사: 선석 점유율, 크레인 생산성
- 화주/포워더: 도착/반출 시각, 비용
- 항만청: 물동량, 환적의 비중
- 해양수상청, 관제(VTS): 입출항 척수, 안전
- 도예/선업체: 작업 횟수, 대기
